```python
nlp = spacy.load("pt_core_news_sm")
```

```python
path = "/content/drive/MyDrive/"
text = "helena.txt"
```

```python
with open(path + text, "r") as textFile:
  textData = textFile.read().replace('\n', ' ')
print(textData[0:1000])
```

```python
doc = nlp(textData[35:1000])
print(doc.text)
for token in doc:
  print(token.text, token.pos_)
```

```python
from spacy import displacy
```

```python
displacy.render(doc, style='dep', jupyter=True)

print("Token -> papel no grafo de dependências -> categoria")
for token in doc:
  print(token.text, "->", token.dep_, "->", token.pos_)
```

```python
def subtree(doc):
  suj, verb, obj = "","",""

  for i, token in enumerate(doc):
    if "ROOT" in token.dep_ and (token.pos_ in 'VERB'):
      verb = token.text
    if "subj" in token.dep_:
      suj = token.text
    if "nmod" in token.dep_ and (token.pos_ in ("NOUN", "PROPN")):
      obj = token.text
    if suj and verb and obj:
      print(suj, ", ", verb, ",", obj)
      suj, verb, obj = "","",""

subtree(doc)
```
