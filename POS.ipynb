pt-core-news refers to a series of pre-trained language models for the Portuguese language, specifically designed for use with the spaCy natural language processing library. These models are optimized for CPU usage and provide a range of functionalities for text analysis.

Key features and capabilities of pt-core-news models include:
* Tokenization: Breaking down text into individual words or sub-word units.
* Part-of-Speech (POS) Tagging: Identifying the grammatical category of each word (e.g., noun, verb, adjective).
* Dependency Parsing: Analyzing the grammatical relationships between words in a sentence.
* Named Entity Recognition (NER): Identifying and classifying named entities such as persons, organizations, and locations.

```python
nlp = spacy.load("pt_core_news_sm")
```

```python
path = "/content/drive/MyDrive/"
text = "helena.txt"
```

```python
with open(path + text, "r") as textFile:
  textData = textFile.read().replace('\n', ' ') # retirando as quebras de linha
print(textData[0:1000]) # lendo só os 1000 primeiros caracteres
indice = textData.find("O conselheiro")
print(indice) # índice onde a história de fato começa
print(textData[indice:indice+1000])
```

```python
doc = nlp(textData[indice:indice+1000])
print(doc.text)
for token in doc:
  print(token.text, token.pos_) # durante a tokenização com o spacy, o POS do token já vem associado como um atributo
```

```python
from spacy import displacy
```

Parsing de dependência entre tokens

```python
displacy.render(doc, style='dep', jupyter=True) # árvore de dependência

print("Token -> papel no grafo de dependências -> categoria")
for token in doc:
  print(token.text, "->", token.dep_, "->", token.pos_)
```

```python
def subtree(doc):
  suj, verb, obj = "","",""

  for i, token in enumerate(doc):
    if "ROOT" in token.dep_ and (token.pos_ in 'VERB'):
      verb = token.text
    if "subj" in token.dep_ and (token.pos_ in ("NOUN", "PROPN")):
      suj = token.text
    if "nmod" in token.dep_ and (token.pos_ in ("NOUN", "PROPN")):
      obj = token.text
    if suj and verb and obj:
      print(suj, ", ", verb, ",", obj)
      suj, verb, obj = "","",""

subtree(doc)
```
