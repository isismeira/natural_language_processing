# Introdução ao NLTK

```python
!pip install nltk==3.8

import nltk
```

```python
from nltk.stem import WordNetLemmatizer # para stemização
from nltk.tokenize import sent_tokenize, word_tokenize # segmentação de sentenças, segmentação de palavras
lemmatizer = WordNetLemmatizer() # objeto que faz a lematização
nltk.download('punkt_tab') # recursos linguísticos
nltk.download('wordnet')
```

```python
texto = 'Não se tratava do mapa encontrado ao pescoço do Billy Bones mas sim de uma cópia muito minuciosa, nomes, altitudes, fundos, com a única exceção das cruzes vermelhas e das notas manuscritas. Por muito que lhe custasse, o Silver teve presença de espírito para esconder a desilusão.'
```

Processo de lematização

```python
nltk.download('punkt')
textolmn = ''
words = sent_tokenize(texto)
for word in words:
  textolmn += lemmatizer.lemmatize(word) + ' '
print(textolmn)
```

Não funciona porque o lematizador do NLTK não funciona bem para português

Já a stemização do NLTK funciona bem

```python
nltk.download('rslp')
from nltk.stem import RSLPStemmer
```

```python
st = RSLPStemmer()
textost = ''
for token in texto.split():
  textost+=st.stem(token) + ' '
print(textost)
```

# Introdução ao Spacy

Funciona melhor para portugês pois tem um lematizador que funciona bem

```python
import spacy
```

```python
!python -m spacy download pt_core_news_sm
sp = spacy.load('pt_core_news_sm')
```

```python
texto = sp('Não se tratava do mapa encontrado ao pescoço do Billy Bones mas sim de uma cópia muito minuciosa, nomes, altitudes, fundos, com a única exceção das cruzes vermelhas e das notas manuscritas. Por muito que lhe custasse, o Silver teve presença de espírito para esconder a desilusão.')
```

```python
textlm = ''
for word in texto:
  textlm+=word.lemma_ + ' '
print(textlm)
```

Diferente do NLTK, o lematizador do Spacy funcionou! Encontrou o radical comum

```python
for token in sp(sentences[1000]): # analisando as sentenças do romance do Machado de Assis
  print("Token -->", token, "Lemma -->", token.lemma_)
```

# Introdução ao Stanza

```python
!pip install stanza
import stanza
```

```python
stanza.download('pt')
nlp = stanza.Pipeline('pt')
```

```python
for sent in nlp(sentences[1000]).sentences: # observando a lematização de ua sentença do romance de Machado
  for word in sent.words:
    print("Token -->", word.text, "Lemma -->", word.lemma)
```
